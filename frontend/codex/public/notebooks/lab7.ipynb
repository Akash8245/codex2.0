```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

```python
Sentences = [
    'I love pandas',
    'Pandas are cute',
    'I am panda'
]
```

```python
Count_Vectorizer = CountVectorizer(ngram_range=(1,1),stop_words='english')
```

```python
Count_Data = Count_Vectorizer.fit_transform(Sentences)
```

```python
DataFrame = pd.DataFrame(Count_Data.toarray(),columns=Count_Vectorizer.get_feature_names_out())
```
