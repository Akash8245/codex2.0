```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Day': np.arange(1, 13),
    'Stock_Price': [100, 102, 101, 105, 110, 115, 120, 118, 122, 125, 130, 128]
}
```

```python

df = pd.DataFrame(data)
```

```python

plt.plot(df['Day'], df['Stock_Price'], marker='o', label='Original Stock Prices')
plt.title('Stock Prices Over Days')
plt.xlabel('Day')
plt.ylabel('Stock Price')
plt.legend()
plt.show()
```

```python
window_size = 3
df['Moving_Average'] = df['Stock_Price'].rolling(window=window_size).mean()

plt.plot(df['Day'], df['Stock_Price'], marker='o', label='Original Stock Prices')
plt.plot(df['Day'], df['Moving_Average'], marker='o', label=f'{window_size}-Day Moving Average')
plt.title('Stock Prices with Moving Average')
plt.xlabel('Day')
plt.ylabel('Stock Price')
plt.legend()
plt.show()
```

```python
forecast = df['Stock_Price'].iloc[-window_size:].mean()
print(f"Forecast for the next day: {forecast:.2f}")
```

```python
new_row = {'Day': 13, 'Stock_Price': np.nan, 'Moving_Average': forecast}
df_forecast = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)

plt.plot(df['Day'], df['Stock_Price'], marker='o', label='Original Stock Prices')
plt.plot(df['Day'], df['Moving_Average'], marker='o', label=f'{window_size}-Day Moving Average')
plt.scatter(13, forecast, color='red', label=f'Forecasted Price ({forecast:.2f})')
plt.title('Stock Prices with Moving Average and Forecast')
plt.xlabel('Day')
plt.ylabel('Stock Price')
plt.legend()
plt.show()
```
