# Step 1:- Import Libraries 

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

# Step 2 :- Give your Own Sentences 

```python
Sentences = [
    'I like Icecream',
    'I love Music',
    'I love Sports'
]
```

# Step 3 :- Vectorizing 

```python
Count_Veactorizer = CountVectorizer(ngram_range=(1,1),stop_words='english')
```

# Step 4:- Transforming

```python
Count_Data = Count_Vectorizer.fit_transform(Sentences)
```

# Step 5 :- Creating a DataFrame 

```python
Data_Frame = pd.DataFrame(Count_Data.toarray(),columns=Count_Vectorizer.get_feature_names_out())
```
