# Step 1 :- Import Libraries 

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder,StandardScaler
```

# Step 2 :-  Import Dataset 

```python
Data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', None],
    'Age': [25, None, 30, 22, 29],
    'Gender': ['Female', 'Male', 'Male', 'Male', 'Female'],
    'Salary': [50000, 60000, None, 52000, 58000],
    'Department': ['HR', 'IT', 'IT', 'HR', 'Finance']
}
```

# Step 3 :- Create a DataFrame 

```python
df = pd.DataFrame(Data)
```

```python
df
```

# Step 4 :- Identify the Null Values in the DataFrame 

```python
df['Age'].fillna(df['Age'].mean()) #Filling Missimg Values inside the Age Column
```

```python
df
```

```python
df['Salary'].fillna(df['Salary'].mean())
```

```python
df
```

# Step 5 :- Handling Categorical Data

```python
Encoder = LabelEncoder()
```

```python
df['Gender'] = Encoder.fit_transform(df['Gender'])
```

```python
df
```

# Step 6 :- Remove Duplicates 

```python
df = df.drop_duplicates()
```

```python
df
```

# Step 7 :- Outlier Treatment 

```python
Q1 = df[['Age','Salary']].quantile(0.25)
Q3 = df[['Age','Salary']].quantile(0.75)
```

```python
IQR = Q3 - Q1
```

```python
IQR
```

```python
Lower_Bound = Q1 - 1.5 * IQR
Upper_Bound = Q3 + 1.5 * IQR
```

```python
for Col in ['Age' , 'Salary']:
    df = df[(df[Col] >= Lower_Bound[Col]) & (df[Col] <= Upper_Bound[Col])]

df
```

# Step 8:- Feature Scaling 

```python
Scaler = StandardScaler()
```

```python
df[['Age','Salary']] = Scaler.fit_transform(df[['Age','Salary']])
```
